```python
def load_preds_single_model(api_key = api_key, model_name = 'str'): 

    preds = mosq.get_predictions(api_key = api_key, model_name= model_name)

    df_info = pd.DataFrame([
        {   "pred": pred, 
            "model_name": pred.model.repository,
            "id": pred.id, 
            "adm_1": pred.adm_1,
            "start": pred.start,
            "end": pred.end,
            "wis": pred.scores.get("wis")
            if pred.scores is not None else None,
            "published": pred.published }
        for pred in preds
    ])

    df_info = df_info.loc[df_info.published == True]

    # removendo as duplicatas mantendo a previsão com o maior id

    df_info = (
            df_info.sort_values("id")
            .drop_duplicates(
                subset=["adm_1", "start", "end"],
                keep="last"
            )
        )


    return df_info 
```
