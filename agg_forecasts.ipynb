```python
import numpy as np
import pandas as pd
from pathlib import Path
from epiweeks import Week
import matplotlib.pyplot as plt 
```

```python
label = 'forecast_dengue_city'

caminho_da_pasta = Path("./predictions/")

padrao_arquivo = f"{label}_*.csv.gz"  # Altere para o seu termo

arquivos_encontrados = list(caminho_da_pasta.glob(padrao_arquivo))

arquivos_encontrados
```

```python
print(len(arquivos_encontrados))
```

```python
arquivos_ignorados = {
    f"{label}_3rd_imdc_emap_epidematicos_prophet.csv.gz",
    f"{label}_3rd_imdc_emap_epidematicos_sarimax_muni.csv.gz",
    f"{label}_3rd_imdc_emap_epidematicos_sarimax_state.csv.gz",
}

arquivos_encontrados = [
    arquivo
    for arquivo in caminho_da_pasta.glob(padrao_arquivo)
    if arquivo.name not in arquivos_ignorados
]

arquivos_encontrados
```

```python
print(len(arquivos_encontrados))
```

```python
list_dfs = []

for arq in arquivos_encontrados:

    df_ = pd.read_csv(arq)
    df_.date = pd.to_datetime(df_.date)
    df_['model'] = str(arq)[33:-7]

    df_ = df_.loc[(df_.date>= pd.to_datetime(Week(2026, 41).startdate())) & (df_.date<= pd.to_datetime(Week(2027, 40).startdate()))]

    list_dfs.append(df_)


df_end = pd.concat(list_dfs, ignore_index = True)

df_end.to_csv(f'./predictions/forecast_all_models_{label[9:]}.csv.gz')


df_end.head()
```
