```python
import pandas as pd
from epiweeks import Week 
```

```python
geo_dengue = [2931350,2933307,2302503,3119401,
              3549805,3541406,1200401,1200203,
              1716109,4113700,4103701,4104808,
              5201405,5102637,5215231]

geo_chik = [2211001,2931350,3143302,3119401,
            1721000,1716109,4104808,4219507,
            5103403,5102637] 
```

```python
df_dengue = pd.read_csv('../../imdc_2026/dengue_update.csv.gz')
df_dengue = df_dengue.loc[df_dengue.geocode.isin(geo_dengue)][['date', 'geocode', 'casos']].rename(columns = {'geocode': 'adm_2'})
df_dengue.head()
```

```python
df_dengue.to_csv('dengue_city.csv.gz', index = False)
```
