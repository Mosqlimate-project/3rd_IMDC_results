```python
import pandas as pd
from epiweeks import Week 
```

```python
df_old = pd.read_csv('chik_state.csv.gz')
df_old.date = pd.to_datetime(df_old.date)
df_old.head()
```

```python
df_old.loc[df_old.date == pd.to_datetime(Week(2023,24).startdate())]
```

```python
df_new = pd.read_csv('../../imdc_2026/chikungunya_update.csv.gz')
df_new = df_new.groupby(['date', 'uf_code'])[['casos']].sum().reset_index().rename(columns = {'uf_code': 'adm_1'})
df_new.date = pd.to_datetime(df_new.date)
df_new.loc[df_new.date>='2023-06-11']
```

```python
df_end = pd.concat([df_old.loc[df_old.date <= pd.to_datetime(Week(2023,24).startdate())], df_new.loc[df_new.date>'2023-06-11']])
```

```python
df_end.to_csv('chik_state.csv.gz', index = False)
```
